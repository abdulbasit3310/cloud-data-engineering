```python
for i in range(10):
    print(i)
    
    if i == 5:
        print("Breaking the loop")
        break # this will break the loop when i is equal to 5
```

```python
# CONTINUE STATEMENT
for i in range(10):
    if i == 5 or i == 7:
        continue # this will skip the iteration when i is equal to 5 or 7
    print(i)
```

```python
while True: # this will create an infinite loop as the condition is always true
    print("Hello World")
```

```python
num = 0

while num < 10:
    print(num)
    num += 1 # this will increment the value of num by 1 in each iteration
```

```python
retries = 0
max_retries = 3
connected = False

while not connected and retries < max_retries:
    print(f"Attempting connection... (attempt {retries + 1})")
    retries += 1
    if retries == 2:
        connected = True

if connected:
    print("Connection established.")
else:
    print("Connection failed after maximum retries.")
```

## NESTED LOOPS

```python
tools = ["AWS","Azure","PySpark","DBT","Snowflake","Databricks"]

for tool in tools:
    print(tool)
    
    for char in tool:
        print(char) # this will print each character of the tool name in a new line

# first loop
# first iteration -- tool = "AWS" --- 2nd loop --- total iterations 3 --- A W S
# second iteration -- tool = "Azure" --- 2nd loop --- total iterations 5 --- A z u r e
# third iteration -- tool = "PySpark" --- 2nd loop --- total iterations 7 --- P y S p a r k
# fourth iteration -- tool = "DBT" --- 2nd loop --- total iterations 3 --- D B T
# fifth iteration -- tool = "Snowflake" --- 2nd loop --- total iterations 9 --- S n o w f l a k e
# sixth iteration -- tool = "Databricks" --- 2nd loop --- total iterations 10 --- D a t a b r i c k s
# total iterations of the inner loop = 3 + 5 + 7 + 3 + 9 + 10 = 37
# total iterations of the outer loop = 6
# total iterations of the nested loop = 37 + 6 = 43
```
